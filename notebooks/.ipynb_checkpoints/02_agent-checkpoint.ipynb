```python
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
db = Chroma(
    persist_directory="C:\\Users\\Nitu28\\startup-policy-copilot\\data\\index",
    embedding_function=embedding_model
)

retriever = db.as_retriever(search_kwargs={"k": 3})
```

```python
def local_policy_agent(question: str):
    docs = retriever.get_relevant_documents(question)
    answer = ""
    for doc in docs:
        answer += doc.page_content[:800] + "\n"  # limit to 800 chars per chunk
        answer += f"source: {doc.metadata['source']}\n\n"
    return answer
```

```python
query = "What is the eligibility criteria for DPIIT startup recognition?"
print(local_policy_agent(query))
```
