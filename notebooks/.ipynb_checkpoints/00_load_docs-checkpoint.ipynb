```python
import os
from PyPDF2 import PdfReader
from tqdm import tqdm
```

```python
RAW_DIR = "C:\\Users\\Nitu28\\startup-policy-copilot\\data\\raw"
CLEAN_DIR = "C:\\Users\\Nitu28\\startup-policy-copilot\\data\\clean"

pdf_files = [f for f in os.listdir(RAW_DIR) if f.endswith(".pdf")]
print(pdf_files)
```

```python
def clean_text(text: str) -> str:
    text = text.replace("\n", " ")            # remove line breaks
    text = ' '.join(text.split())              # collapse multiple spaces
    return text
```

```python
for pdf_name in tqdm(pdf_files):
    pdf_path = os.path.join(RAW_DIR, pdf_name)
    reader = PdfReader(pdf_path)
    all_text = ""
    for page in reader.pages:
        page_text = page.extract_text()
        if page_text:
            cleaned = clean_text(page_text)
            all_text += cleaned + "\n"
    # save to .txt file
    clean_filename = pdf_name.replace(".pdf", ".txt")
    clean_path = os.path.join(CLEAN_DIR, clean_filename)
    with open(clean_path, "w", encoding="utf-8") as f:
        f.write(all_text)
    print(f"Saved cleaned text to {clean_filename}")
```

```python
import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
```

```python
from pdf2image import convert_from_path

def ocr_pdf(pdf_path):
    images = convert_from_path(pdf_path)
    text = ""
    for img in images:
        page_text = pytesseract.image_to_string(img)
        text += page_text + "\n"
    return text
```

```python
ocr_pdfs = ["dpiit_startup_recognition.pdf", 
            "sisfs_portal_faqs.pdf", 
            "startup_action_plan_status.pdf"]

for pdf_name in ocr_pdfs:
    pdf_path = os.path.join(RAW_DIR, pdf_name)
    text = ocr_pdf(pdf_path)
    cleaned = clean_text(text)
    clean_filename = pdf_name.replace(".pdf", ".txt")
    clean_path = os.path.join(CLEAN_DIR, clean_filename)
    with open(clean_path, "w", encoding="utf-8") as f:
        f.write(cleaned)
    print(f"OCR complete → saved to {clean_filename}")
```
