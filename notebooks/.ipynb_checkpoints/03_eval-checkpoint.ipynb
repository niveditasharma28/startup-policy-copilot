```python
import json

with open("C:\\Users\\Nitu28\\startup-policy-copilot\\eval\\eval_set.json", encoding="utf-8") as f:
    eval_qs = json.load(f)

from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

db = Chroma(
    persist_directory="C:\\Users\\Nitu28\\startup-policy-copilot\\data\\index",
    embedding_function=embedding_model
)

retriever = db.as_retriever(search_kwargs={"k": 3})

def local_policy_agent(question: str):
    docs = retriever.get_relevant_documents(question)
    answer = ""
    for doc in docs:
        answer += doc.page_content[:800] + "\n"
        answer += f"source: {doc.metadata['source']}\n\n"
    return answer

from difflib import SequenceMatcher

def similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()

scores = []

for item in eval_qs:
    question = item["question"]
    expected_list = item["expected"]  # this is a list of acceptable phrases
    answer = local_policy_agent(question).lower()
    
    matched = False
    for expected in expected_list:
        expected = expected.lower()
        sim = similarity(expected, answer)
        if sim > 0.4:
            matched = True
            break
    
    score = 1 if matched else 0
    scores.append(score)
    
    print(f"Q: {question}")
    print(f"Got (first 500 chars): {answer[:500]}")
    print(f"Matched: {matched}\n")

accuracy = sum(scores) / len(scores)
print(f"Overall Evaluation Accuracy: {accuracy*100:.2f}%")

```
